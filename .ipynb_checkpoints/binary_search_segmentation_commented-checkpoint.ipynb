## 2. Algorithm

### 2.1 Preprocessing and features

```python
# ---------------- Tham số cố định theo đề bài (KHÔNG được tối ưu) ----------------
FRAME_MS = 25.0     # độ dài khung (ms)
HOP_MS = 10.0       # bước dịch khung (ms)
MIN_SIL_MS = 200.0  # độ dài tối thiểu của một đoạn im lặng (ms)
EPS_LOG = 1e-6      # hằng số nhỏ để tránh log(0)


def preprocess(x):
    """Tiền xử lý: loại thành phần một chiều (DC) rồi chuẩn hóa biên độ đỉnh về 1.

    Tham số:
        x (array-like): tín hiệu gốc (các mẫu thực).
    Trả về:
        ndarray float64: tín hiệu đã bỏ DC và chia cho biên độ đỉnh;
        tín hiệu toàn 0 được giữ nguyên.
    """
    # Đổi sang float64 rồi trừ giá trị trung bình để khử DC
    x = np.asarray(x, dtype=np.float64)
    x = x - np.mean(x)
    # Lấy biên độ đỉnh (0 nếu tín hiệu rỗng) và chuẩn hóa về đoạn [-1, 1]
    peak = np.max(np.abs(x)) if len(x) else 0.0
    return x / peak if peak > 0 else x


def frame_params(fs):
    """Đổi độ dài khung 25 ms và bước dịch 10 ms sang số mẫu theo tần số lấy mẫu của file.

    Tham số:
        fs (int): tần số lấy mẫu (Hz).
    Trả về:
        (frame_len, hop): độ dài khung và bước dịch, tính bằng số mẫu.
    """
    return int(round(FRAME_MS * fs / 1000.0)), int(round(HOP_MS * fs / 1000.0))


def num_frames(n_samples, frame_len, hop):
    """Tính số khung của tín hiệu (khung cuối nằm trọn trong tín hiệu; tối thiểu 1 khung).

    Tham số:
        n_samples (int): số mẫu của tín hiệu.
        frame_len (int): độ dài khung (mẫu).
        hop (int): bước dịch khung (mẫu).
    Trả về:
        int: số khung.
    """
    if n_samples < frame_len:
        return 1
    return 1 + (n_samples - frame_len) // hop


def frame_centers(n_frames, frame_len, hop, fs):
    """Tính thời điểm tâm của từng khung.

    Tham số:
        n_frames (int): số khung.
        frame_len (int): độ dài khung (mẫu).
        hop (int): bước dịch khung (mẫu).
        fs (int): tần số lấy mẫu (Hz).
    Trả về:
        ndarray: thời điểm tâm mỗi khung (giây), độ dài n_frames.
    """
    return (np.arange(n_frames) * hop + frame_len / 2.0) / fs


def frame_features(x, fs, kind="MA"):
    """Tính đặc trưng năng lượng cho từng khung bằng tổng tích lũy (không dựng ma trận khung).

    Tham số:
        x (ndarray): tín hiệu đã tiền xử lý.
        fs (int): tần số lấy mẫu (Hz).
        kind (str): 'MA' = mean(|x|), 'RMS' = sqrt(mean(x^2)), 'STE' = sum(x^2).
    Trả về:
        ndarray: giá trị đặc trưng của mỗi khung.
    """
    # Lấy thông số khung; nếu tín hiệu ngắn hơn một khung thì đệm thêm số 0
    frame_len, hop = frame_params(fs)
    if len(x) < frame_len:
        x = np.concatenate((x, np.zeros(frame_len - len(x))))
    # Xác định chỉ số bắt đầu/kết thúc của mỗi khung
    nf = num_frames(len(x), frame_len, hop)
    starts = np.arange(nf) * hop
    ends = starts + frame_len

    # MA: trung bình biên độ tuyệt đối, tính nhanh qua tổng tích lũy của |x|
    if kind == "MA":
        c = np.concatenate(([0.0], np.cumsum(np.abs(x))))
        return (c[ends] - c[starts]) / frame_len

    # STE / RMS: tổng bình phương mỗi khung qua tổng tích lũy của x^2
    c2 = np.concatenate(([0.0], np.cumsum(x * x)))
    ste = np.maximum(c2[ends] - c2[starts], 0.0)
    if kind == "STE":
        return ste
    if kind == "RMS":
        return np.sqrt(ste / frame_len)
    raise ValueError("kind must be MA, RMS or STE")


def moving_average(a, length):
    """Làm mượt chuỗi bằng trung bình trượt (cửa sổ lẻ, hai đầu được nhân bản giá trị biên).

    Tham số:
        a (ndarray): chuỗi đầu vào.
        length (int): độ dài cửa sổ (nên là số lẻ để đầu ra cùng độ dài với đầu vào).
    Trả về:
        ndarray: chuỗi đã làm mượt (length <= 1 thì trả về nguyên chuỗi gốc).
    """
    if length <= 1:
        return a
    # Đệm h phần tử giá trị biên vào hai đầu để không mất mẫu ở rìa
    h = length // 2
    p = np.concatenate((np.full(h, a[0]), a, np.full(h, a[-1])))
    # Tính trung bình trượt bằng tổng tích lũy
    c = np.concatenate(([0.0], np.cumsum(p)))
    return (c[length:] - c[:-length]) / length


def compute_feature_track(x_norm, fs, kind, use_log, smooth_len):
    """Tính đường đặc trưng cuối cùng: đặc trưng từng khung -> làm mượt -> (tùy chọn) log10.

    Tham số:
        x_norm (ndarray): tín hiệu đã chuẩn hóa.
        fs (int): tần số lấy mẫu (Hz).
        kind (str): loại đặc trưng ('MA', 'RMS', 'STE').
        use_log (bool): True thì lấy log10 (có cộng EPS_LOG).
        smooth_len (int): độ dài cửa sổ làm mượt (số khung).
    Trả về:
        ndarray: giá trị đặc trưng của mỗi khung sau xử lý.
    """
    feat = moving_average(frame_features(x_norm, fs, kind), smooth_len)
    if use_log:
        feat = np.log10(feat + EPS_LOG)
    return feat
```

### 2.2 Threshold search by binary search

```python
def find_overlap_region(S, Q, trim=1.0):
    """Tìm vùng chồng lấn giữa đặc trưng tiếng nói và im lặng.

    Tham số:
        S (ndarray): đặc trưng các khung tiếng nói.
        Q (ndarray): đặc trưng các khung im lặng.
        trim (float): phần trăm bị cắt ở mỗi đầu để loại giá trị ngoại lai.
    Trả về:
        (lo, hi): lo = phân vị trim% của S, hi = phân vị (100 - trim)% của Q.
    """
    return np.percentile(S, trim), np.percentile(Q, 100.0 - trim)


def confusion_error(T, f, g):
    """Tính hàm sai số F(T) = (tỷ lệ khung im lặng > T) - (tỷ lệ khung tiếng nói <= T).
    F giảm đơn điệu theo T nên có thể tìm nghiệm F(T)=0 bằng chia đôi.

    Tham số:
        T (float): ngưỡng đang xét.
        f (ndarray): đặc trưng khung tiếng nói trong vùng chồng lấn.
        g (ndarray): đặc trưng khung im lặng trong vùng chồng lấn.
    Trả về:
        (F, (n_f_below, n_g_above)): giá trị F(T) và hai số đếm dùng để phát hiện hội tụ.
    """
    n_f_below = int(np.sum(f <= T))   # số khung tiếng nói bị xếp nhầm vào im lặng
    n_g_above = int(np.sum(g > T))    # số khung im lặng bị xếp nhầm vào tiếng nói
    return n_g_above / len(g) - n_f_below / len(f), (n_f_below, n_g_above)


def binary_search_threshold(f, g, lo, hi, eps_rel=1e-4, max_iter=60, patience=3):
    """Tìm ngưỡng T sao cho F(T) = 0 bằng tìm kiếm nhị phân trên đoạn [lo, hi].

    Tham số:
        f, g (ndarray): đặc trưng tiếng nói / im lặng trong vùng chồng lấn.
        lo, hi (float): hai đầu đoạn tìm kiếm.
        eps_rel (float): độ rộng đoạn dừng, tính theo tỷ lệ của (hi - lo).
        max_iter (int): số vòng lặp tối đa.
        patience (int): số vòng liên tiếp số đếm không đổi thì dừng sớm.
    Trả về:
        float: ngưỡng T tìm được (trung điểm đoạn cuối cùng).
    """
    # Khởi tạo đoạn tìm kiếm, sai số dừng và ngưỡng đầu tiên ở giữa đoạn
    t_min, t_max = lo, hi
    eps = eps_rel * (hi - lo)
    T = 0.5 * (t_min + t_max)
    last, same = None, 0

    for _ in range(max_iter):
        # Tính sai số tại T: F > 0 nghĩa là T quá thấp (nhiều khung im lặng vượt T)
        F, counts = confusion_error(T, f, g)
        if F > 0:
            t_min = T
        else:
            t_max = T
        # Điều kiện dừng: đoạn đủ hẹp, hoặc số đếm không đổi trong `patience` vòng
        if t_max - t_min < eps:
            break
        same = same + 1 if counts == last else 0
        last = counts
        if same >= patience:
            break
        T = 0.5 * (t_min + t_max)
    return 0.5 * (t_min + t_max)


def train_threshold(S, Q, trim=1.0):
    """Tìm ngưỡng T từ đặc trưng khung tiếng nói (S) và khung im lặng (Q) của tập huấn luyện.

    Tham số:
        S (ndarray): đặc trưng các khung tiếng nói.
        Q (ndarray): đặc trưng các khung im lặng.
        trim (float): phần trăm cắt ngoại lai khi tìm vùng chồng lấn.
    Trả về:
        (T, sep): T là ngưỡng; sep = mean(S) - mean(Q) dùng để đặt độ rộng trễ (hysteresis).
    """
    # Tính độ tách giữa hai lớp và vùng chồng lấn của chúng
    S, Q = np.asarray(S), np.asarray(Q)
    sep = float(np.mean(S) - np.mean(Q))
    lo, hi = find_overlap_region(S, Q, trim)
    # Hai lớp không chồng lấn: lấy trung điểm khoảng trống làm ngưỡng
    if lo >= hi:
        return 0.5 * (lo + hi), sep
    # Giữ lại các khung nằm trong vùng chồng lấn; nếu một lớp rỗng thì dùng trung điểm
    f = S[(S >= lo) & (S <= hi)]
    g = Q[(Q >= lo) & (Q <= hi)]
    if len(f) == 0 or len(g) == 0:
        return 0.5 * (lo + hi), sep
    # Tìm ngưỡng bằng tìm kiếm nhị phân trên vùng chồng lấn
    return binary_search_threshold(f, g, lo, hi), sep
```

### 2.3 Frame labeling, post-processing and boundary conversion

```python
def classify_frames(feat, T, sep, hyst):
    """Gán nhãn từng khung: True = tiếng nói, False = im lặng.

    Tham số:
        feat (ndarray): đường đặc trưng của các khung.
        T (float): ngưỡng quyết định.
        sep (float): độ tách hai lớp (mean(S) - mean(Q)), dùng nhân với hyst.
        hyst (float): hệ số trễ; hyst = 0 dùng một ngưỡng, hyst > 0 dùng hai ngưỡng
            (bật tại T + hyst*sep, tắt tại T - hyst*sep).
    Trả về:
        ndarray bool: mặt nạ tiếng nói của các khung.
    """
    # Trường hợp một ngưỡng: so sánh trực tiếp với T
    if hyst <= 0:
        return feat > T
    # Trường hợp hai ngưỡng: tính ngưỡng bật (t_on) và ngưỡng tắt (t_off)
    t_on, t_off = T + hyst * sep, T - hyst * sep
    mask = np.zeros(len(feat), dtype=bool)
    state = False
    # Duyệt tuần tự: đang nói thì chỉ tắt khi < t_off, đang im thì chỉ bật khi > t_on
    for k, v in enumerate(feat.tolist()):
        if state:
            if v < t_off:
                state = False
        elif v > t_on:
            state = True
        mask[k] = state
    return mask


def dilate(mask, e):
    """Mở rộng các vùng tiếng nói thêm e khung về mỗi phía.

    Tham số:
        mask (ndarray bool): mặt nạ tiếng nói.
        e (int): số khung mở rộng mỗi phía (e <= 0 thì không đổi).
    Trả về:
        ndarray bool: mặt nạ sau khi mở rộng.
    """
    if e <= 0:
        return mask
    # Tổng tích lũy của mặt nạ để đếm nhanh số khung nói trong cửa sổ [i-e, i+e]
    n = len(mask)
    c = np.concatenate(([0], np.cumsum(mask.astype(np.int64))))
    lo = np.maximum(np.arange(n) - e, 0)
    hi = np.minimum(np.arange(n) + e + 1, n)
    # Khung được coi là tiếng nói nếu trong cửa sổ có ít nhất một khung nói
    return (c[hi] - c[lo]) > 0


def trim_speech(mask, trim_on, trim_off):
    """Dời điểm bắt đầu nói muộn đi trim_on khung và điểm kết thúc nói sớm lại trim_off khung.
    Giá trị âm có nghĩa là mở rộng vùng tiếng nói. Biên sát đầu/cuối file không bị dời.

    Tham số:
        mask (ndarray bool): mặt nạ tiếng nói.
        trim_on (int): số khung dời điểm bắt đầu nói (dương = muộn hơn).
        trim_off (int): số khung dời điểm kết thúc nói (dương = sớm hơn).
    Trả về:
        ndarray bool: mặt nạ sau khi dời biên.
    """
    if trim_on == 0 and trim_off == 0:
        return mask
    n = len(mask)
    out = np.zeros(n, dtype=bool)
    # Với mỗi đoạn tiếng nói: dời đầu đoạn và cuối đoạn rồi chặn trong [0, n]
    for s, e, v in get_runs(mask):
        if not v:
            continue
        s2 = s if s == 0 else min(max(s + trim_on, 0), n)
        e2 = e if e == n else min(max(e - trim_off, 0), n)
        # Chỉ giữ lại đoạn nếu sau khi dời vẫn còn độ dài dương
        if e2 > s2:
            out[s2:e2] = True
    return out


def get_runs(mask):
    """Tách mặt nạ thành các đoạn liên tiếp cùng nhãn.

    Tham số:
        mask (ndarray bool): mặt nạ nhãn theo khung.
    Trả về:
        list[(first_frame, last_frame_exclusive, label)]: danh sách đoạn, mỗi đoạn gồm
        khung đầu, khung cuối (không tính) và nhãn bool.
    """
    # Tìm các vị trí nhãn đổi giá trị để xác định đầu/cuối từng đoạn
    m = mask.astype(np.int8)
    change = np.nonzero(m[1:] != m[:-1])[0] + 1
    starts = np.concatenate(([0], change))
    ends = np.concatenate((change, [len(m)]))
    return [(int(s), int(e), bool(m[s])) for s, e in zip(starts, ends)]


def remove_short_runs(mask, value, min_frames, include_edges=True):
    """Đảo nhãn các đoạn mang nhãn `value` mà ngắn hơn min_frames khung.

    Tham số:
        mask (ndarray bool): mặt nạ nhãn theo khung.
        value (bool): nhãn của các đoạn cần kiểm tra (True = nói, False = im lặng).
        min_frames (int): độ dài tối thiểu (khung) để giữ đoạn.
        include_edges (bool): False thì bỏ qua đoạn nằm ở đầu/cuối file.
    Trả về:
        ndarray bool: mặt nạ sau khi loại các đoạn quá ngắn.
    """
    out = mask.copy()
    runs = get_runs(mask)
    for i, (s, e, v) in enumerate(runs):
        # Bỏ qua đoạn khác nhãn hoặc đã đủ dài
        if v != value or (e - s) >= min_frames:
            continue
        # Tùy chọn: không xử lý đoạn ở hai đầu file
        if (i == 0 or i == len(runs) - 1) and not include_edges:
            continue
        # Đảo nhãn đoạn quá ngắn
        out[s:e] = not value
    return out


def mask_to_segments(mask, fs, n_samples):
    """Đổi mặt nạ khung thành danh sách biên và danh sách đoạn theo thời gian.

    Tham số:
        mask (ndarray bool): mặt nạ tiếng nói theo khung.
        fs (int): tần số lấy mẫu (Hz).
        n_samples (int): số mẫu của tín hiệu gốc (để tính độ dài file).
    Trả về:
        (boundaries, segments): boundaries là danh sách thời điểm biên (giây);
        segments là danh sách (t_bắt_đầu, t_kết_thúc, 'speech' | 'sil').
    """
    # Lấy các đoạn liên tiếp và độ dài file (giây)
    frame_len, hop = frame_params(fs)
    dur = n_samples / fs
    runs = get_runs(mask)
    # Biên nằm giữa tâm hai khung kề nhau tại chỗ nhãn đổi
    boundaries = [(s * hop - hop / 2.0 + frame_len / 2.0) / fs for s, _, _ in runs[1:]]
    # Ghép thành các đoạn [edges[i], edges[i+1]] kèm nhãn
    edges = [0.0] + boundaries + [dur]
    segments = [(edges[i], edges[i + 1], "speech" if runs[i][2] else "sil")
                for i in range(len(runs))]
    return boundaries, segments


def predict_file(x, fs, params):
    """Toàn bộ quy trình phân đoạn cho một tín hiệu (xem BinarySearchAlgorithm.predict).

    Tham số:
        x (ndarray): tín hiệu gốc.
        fs (int): tần số lấy mẫu (Hz).
        params (dict): tham số đã huấn luyện (kind, use_log, smooth_len, T, sep, hyst,
            trim_on, trim_off, min_speech_ms; expand nếu có).
    Trả về:
        dict gồm: xn (tín hiệu chuẩn hóa), feat (đặc trưng), mask_raw (nhãn khung thô),
        mask_exp (sau dời biên), mask (nhãn cuối cùng), boundaries (giây), segments.
    """
    # Bước 1: tiền xử lý, tính đặc trưng và gán nhãn khung theo ngưỡng
    xn = preprocess(x)
    feat = compute_feature_track(xn, fs, params["kind"], params["use_log"],
                                 params["smooth_len"])
    mask_raw = classify_frames(feat, params["T"], params["sep"], params["hyst"])

    # Bước 2: hậu xử lý - dời biên (trim), bỏ đoạn nói quá ngắn,
    # rồi lấp các đoạn im lặng ngắn hơn 200 ms (quy tắc của đề bài)
    mask_exp = dilate(mask_raw, int(params.get("expand", 0)))      # giữ để tương thích tham số cũ
    mask_exp = trim_speech(mask_exp, int(params.get("trim_on", 0)),
                           int(params.get("trim_off", 0)))
    min_sp = int(np.ceil(params["min_speech_ms"] / HOP_MS))
    mask_exp = remove_short_runs(mask_exp, True, min_sp)
    min_sil = int(np.ceil(MIN_SIL_MS / HOP_MS))
    mask = remove_short_runs(mask_exp, False, min_sil)

    # Bước 3: đổi mặt nạ cuối cùng thành biên (giây) và các đoạn
    boundaries, segments = mask_to_segments(mask, fs, len(x))
    return {"xn": xn, "feat": feat, "mask_raw": mask_raw, "mask_exp": mask_exp,
            "mask": mask, "boundaries": boundaries, "segments": segments}
```

## 3. Data loading (WAV + LAB)

```python
# Lưới siêu tham số (giữ nhỏ cho nhanh; có thể điều chỉnh)
GRID = {
    "kind": ["MA", "RMS"],            # loại đặc trưng năng lượng
    "use_log": [False, True],         # có lấy log10 hay không
    "smooth_len": [1, 3, 5, 9],       # độ dài cửa sổ làm mượt (khung)
    "hyst": [0.0, 0.15],              # hệ số trễ (0 = một ngưỡng)
    "min_speech_ms": [0, 50, 100],    # độ dài tối thiểu đoạn nói (ms)
}
# Dời biên bắt đầu/kết thúc nói (đơn vị khung 10 ms; âm = mở rộng), tinh chỉnh sau lưới chính
TRIM_RANGE = range(-2, 7)   # -20 ms ... +60 ms

# Theo README của đề: sil = im lặng; v (hữu thanh) và uv (vô thanh) đều tính là TIẾNG NÓI
SIL_LABELS = {"sil"}


# Đọc dữ liệu
# ---------------------------------------------------------------------
def read_wav(path):
    """Đọc file WAV PCM 8/16/24/32 bit và đổi sang tín hiệu mono trong [-1, 1].

    Tham số:
        path (str): đường dẫn file .wav.
    Trả về:
        (x, fs): x là mảng float64 mono trong [-1, 1], fs là tần số lấy mẫu (Hz).
    """
    # Mở file và đọc thông số (số kênh, độ rộng mẫu, tần số lấy mẫu) cùng dữ liệu thô
    with wave.open(path, "rb") as w:
        n_ch, sw = w.getnchannels(), w.getsampwidth()
        fs, n = w.getframerate(), w.getnframes()
        raw = w.readframes(n)
    # Giải mã dữ liệu thô theo độ rộng mẫu và chuẩn hóa về [-1, 1]
    if sw == 1:
        x = (np.frombuffer(raw, dtype=np.uint8).astype(np.float64) - 128.0) / 128.0
    elif sw == 2:
        x = np.frombuffer(raw, dtype="<i2").astype(np.float64) / 32768.0
    elif sw == 3:
        b = np.frombuffer(raw, dtype=np.uint8).reshape(-1, 3).astype(np.int32)
        v = b[:, 0] | (b[:, 1] << 8) | (b[:, 2] << 16)
        v = np.where(v >= 2 ** 23, v - 2 ** 24, v)
        x = v.astype(np.float64) / 2 ** 23
    elif sw == 4:
        x = np.frombuffer(raw, dtype="<i4").astype(np.float64) / 2 ** 31
    else:
        raise ValueError("Unsupported sample width: %d bytes" % sw)
    # Nhiều kênh: lấy trung bình các kênh để thành mono
    if n_ch > 1:
        x = x[: len(x) // n_ch * n_ch].reshape(-1, n_ch).mean(axis=1)
    return x, fs


def read_lab(path):
    """Đọc file nhãn .lab (t_bắt_đầu t_kết_thúc nhãn); bỏ qua các dòng F0mean/F0std.

    Tham số:
        path (str): đường dẫn file .lab.
    Trả về:
        list[(t_bắt_đầu, t_kết_thúc, is_speech)]: các đoạn nhãn, is_speech=False nếu là 'sil'.
    """
    segs = []
    with open(path, "r", encoding="utf-8", errors="ignore") as fh:
        for line in fh:
            # Bỏ dòng thiếu cột hoặc dòng có hai cột đầu không phải số (F0mean, F0std...)
            tok = line.strip().split()
            if len(tok) < 3:
                continue
            try:
                s, e = float(tok[0]), float(tok[1])
            except ValueError:
                continue
            # Lưu đoạn kèm cờ tiếng nói (nhãn khác 'sil' đều là tiếng nói)
            segs.append((s, e, tok[2].strip().lower() not in SIL_LABELS))
    return segs


def lab_boundaries(segs):
    """Lấy biên chuẩn (ground truth) = các điểm đổi nhãn giữa tiếng nói và im lặng.
    Việc đổi giữa v và uv không được tính là biên.

    Tham số:
        segs (list): kết quả của read_lab.
    Trả về:
        list[float]: thời điểm các biên chuẩn (giây).
    """
    return [segs[i][0] for i in range(1, len(segs)) if segs[i][2] != segs[i - 1][2]]


def lab_frame_mask(segs, centers):
    """Tạo nhãn chuẩn cho từng khung (True = tiếng nói) tại tâm khung.

    Tham số:
        segs (list): kết quả của read_lab.
        centers (ndarray): thời điểm tâm các khung (giây).
    Trả về:
        ndarray bool: nhãn chuẩn của mỗi khung.
    """
    mask = np.zeros(len(centers), dtype=bool)
    # Khung có tâm nằm trong một đoạn tiếng nói thì được gán True
    for s, e, is_sp in segs:
        if is_sp:
            mask[(centers >= s) & (centers < e)] = True
    return mask


def boundary_errors(gt_b, pr_b, cap_ms):
    """Tính khoảng cách (ms) từ mỗi biên chuẩn tới biên dự đoán gần nhất, và ngược lại.

    Tham số:
        gt_b (list): biên chuẩn (giây).
        pr_b (list): biên dự đoán (giây).
        cap_ms (float): sai số gán khi một bên không có biên nào (ms).
    Trả về:
        (e_gt, e_pr): e_gt là sai số từ mỗi biên chuẩn tới biên dự đoán gần nhất,
        e_pr là sai số từ mỗi biên dự đoán tới biên chuẩn gần nhất (ms).
    """
    gt, pr = np.asarray(gt_b, float), np.asarray(pr_b, float)
    # Trường hợp một bên rỗng: gán sai số tối đa cap_ms
    if len(gt) == 0:
        return np.array([]), np.full(len(pr), cap_ms)
    if len(pr) == 0:
        return np.full(len(gt), cap_ms), np.array([])
    # Ma trận khoảng cách mọi cặp (chuẩn, dự đoán); lấy min theo từng chiều
    d = np.abs(gt[:, None] - pr[None, :]) * 1000.0
    return d.min(axis=1), d.min(axis=0)


def load_one(wav_path, lab_path):
    """Nạp một cặp wav + lab và tính sẵn các đại lượng dùng khi huấn luyện.

    Tham số:
        wav_path (str): đường dẫn file .wav.
        lab_path (str): đường dẫn file .lab tương ứng.
    Trả về:
        dict gồm: name, x (gốc), xn (chuẩn hóa), fs, dur (giây),
        gt_b (biên chuẩn), gt_mask (nhãn chuẩn theo khung).
    """
    # Đọc tín hiệu và nhãn
    x, fs = read_wav(wav_path)
    dur = len(x) / fs
    segs = read_lab(lab_path)
    # Tính tâm các khung để dựng nhãn chuẩn theo khung
    frame_len, hop = frame_params(fs)
    nf = num_frames(len(x), frame_len, hop)
    centers = frame_centers(nf, frame_len, hop, fs)
    return {"name": os.path.splitext(os.path.basename(wav_path))[0],
            "x": x, "xn": preprocess(x), "fs": fs, "dur": dur,
            "gt_b": lab_boundaries(segs), "gt_mask": lab_frame_mask(segs, centers)}


def load_dataset(training_files, training_labs):
    """Nạp một danh sách các cặp (wav, lab). Hai danh sách phải cùng độ dài và cùng thứ tự.

    Tham số:
        training_files (list[str]): các đường dẫn .wav.
        training_labs (list[str]): các đường dẫn .lab tương ứng.
    Trả về:
        list[dict]: kết quả load_one cho từng cặp.
    """
    if len(training_files) != len(training_labs):
        raise ValueError("Number of WAV files (%d) differs from number of LAB files (%d)"
                         % (len(training_files), len(training_labs)))
    return [load_one(w, l) for w, l in zip(training_files, training_labs)]


def find_pairs(folder):
    """Tìm các cặp *.wav + *.lab cùng tên trong một thư mục (file thiếu .lab bị bỏ qua).

    Tham số:
        folder (str): đường dẫn thư mục dữ liệu.
    Trả về:
        (files, labs): danh sách đường dẫn .wav và .lab tương ứng (cùng thứ tự).
    """
    # Kiểm tra thư mục tồn tại rồi liệt kê các file .wav (không phân biệt hoa/thường)
    if not os.path.isdir(folder):
        raise SystemExit("ERROR: data folder not found: %s" % os.path.abspath(folder))
    wavs = sorted(set(glob.glob(os.path.join(folder, "*.wav"))
                      + glob.glob(os.path.join(folder, "*.WAV"))))
    files, labs = [], []
    # Chỉ giữ file .wav có file .lab cùng tên
    for w in wavs:
        lab = os.path.splitext(w)[0] + ".lab"
        if os.path.exists(lab):
            files.append(w)
            labs.append(lab)
        else:
            print("  [skipped] missing .lab file for", os.path.basename(w))
    if not files:
        raise SystemExit("ERROR: no .wav + .lab pairs found in %s" % folder)
    return files, labs
```

## 4. Training (hyperparameter grid + threshold search by binary search)

```python
# Huấn luyện
# ---------------------------------------------------------------------
def pooled_class_features(data, tracks):
    """Gộp đặc trưng khung của mọi file thành S (tiếng nói) và Q (im lặng) theo nhãn chuẩn.

    Tham số:
        data (list[dict]): dữ liệu từ load_dataset (cần khóa 'gt_mask').
        tracks (list[ndarray]): đường đặc trưng của từng file (cùng thứ tự với data).
    Trả về:
        (S, Q): S là đặc trưng các khung tiếng nói, Q là đặc trưng các khung im lặng.
    """
    S = np.concatenate([t[d["gt_mask"][: len(t)]] for d, t in zip(data, tracks)])
    Q = np.concatenate([t[~d["gt_mask"][: len(t)]] for d, t in zip(data, tracks)])
    return S, Q


def objective(data, params):
    """Hàm mục tiêu (ms, càng nhỏ càng tốt) = trung bình của MAE(chuẩn -> dự đoán)
    và MAE(dự đoán -> chuẩn). Thành phần thứ hai phạt việc sinh biên thừa.

    Tham số:
        data (list[dict]): dữ liệu huấn luyện từ load_dataset.
        params (dict): bộ tham số cần đánh giá.
    Trả về:
        float: điểm số (ms).
    """
    # Dự đoán từng file và gom sai số hai chiều
    e_gt, e_pr = [], []
    for d in data:
        res = predict_file(d["x"], d["fs"], params)
        a, b = boundary_errors(d["gt_b"], res["boundaries"], d["dur"] * 1000.0)
        e_gt.append(a)
        e_pr.append(b)
    # Tính MAE theo từng chiều rồi lấy trung bình
    e_gt, e_pr = np.concatenate(e_gt), np.concatenate(e_pr)
    m1 = np.mean(e_gt) if len(e_gt) else 0.0
    m2 = np.mean(e_pr) if len(e_pr) else 0.0
    return 0.5 * (m1 + m2)


def grid_search(data, verbose=True):
    """Tìm bộ tham số tốt nhất trên lưới siêu tham số GRID.

    Tham số:
        data (list[dict]): dữ liệu huấn luyện từ load_dataset.
        verbose (bool): True thì in tiến trình.
    Trả về:
        (best_params, best_score_ms): bộ tham số tốt nhất và điểm huấn luyện tương ứng.
    """
    best, best_score = None, float("inf")

    # Vòng ngoài: với mỗi (kind, use_log, smooth_len) tính đặc trưng và tìm ngưỡng T
    for kind, use_log, sm in itertools.product(GRID["kind"], GRID["use_log"],
                                               GRID["smooth_len"]):
        tracks = [compute_feature_track(d["xn"], d["fs"], kind, use_log, sm) for d in data]
        S, Q = pooled_class_features(data, tracks)
        T, sep = train_threshold(S, Q)          # tìm kiếm nhị phân trên đặc trưng đã gộp

        # Vòng trong: thử các tham số hậu xử lý, giữ bộ có điểm thấp nhất
        for hy, ms in itertools.product(GRID["hyst"], GRID["min_speech_ms"]):
            params = {"kind": kind, "use_log": use_log, "smooth_len": sm,
                      "T": float(T), "sep": float(sep), "hyst": hy,
                      "trim_on": 0, "trim_off": 0, "min_speech_ms": ms}
            score = objective(data, params)
            if score < best_score:
                best, best_score = params, score

        # In tiến trình sau mỗi tổ hợp (kind, use_log, smooth_len)
        if verbose:
            print("  %-3s log=%-5s smooth=%d  T=%.5f | best so far: %.1f ms"
                  % (kind, use_log, sm, T, best_score))
    return best, best_score


def refine_trim(data, params, verbose=True):
    """Sau lưới chính: tìm mức dời biên tốt nhất (trim_on, trim_off) trên tập huấn luyện.

    Tham số:
        data (list[dict]): dữ liệu huấn luyện từ load_dataset.
        params (dict): bộ tham số tốt nhất từ grid_search.
        verbose (bool): True thì in kết quả.
    Trả về:
        (best, best_score): bộ tham số đã cập nhật trim và điểm huấn luyện tương ứng.
    """
    # Bắt đầu từ bộ tham số hiện tại (trim = 0)
    best, best_score = dict(params), objective(data, params)
    # Thử mọi cặp (trim_on, trim_off) trong TRIM_RANGE, giữ cặp cho điểm thấp nhất
    for ton, toff in itertools.product(TRIM_RANGE, TRIM_RANGE):
        p = dict(params, trim_on=ton, trim_off=toff)
        sc = objective(data, p)
        if sc < best_score:
            best, best_score = p, sc
    if verbose:
        print("Refined trim: trim_on=%d trim_off=%d | training score %.2f ms"
              % (best["trim_on"], best["trim_off"], best_score))
    return best, best_score


def train_binary_search(training_files, training_labs, verbose=True):
    """Huấn luyện từ danh sách đường dẫn WAV và LAB của tập huấn luyện.

    Tham số:
        training_files (list[str]): các đường dẫn .wav.
        training_labs (list[str]): các đường dẫn .lab tương ứng.
        verbose (bool): True thì in tiến trình và kết quả.
    Trả về:
        (params, train_score_ms): params là dict có thể lưu JSON gồm kind, use_log,
        smooth_len, T, sep, hyst, trim_on, trim_off, min_speech_ms; kèm điểm huấn luyện (ms).
    """
    # Nạp dữ liệu huấn luyện, báo lỗi nếu rỗng
    data = load_dataset(training_files, training_labs)
    if not data:
        raise ValueError("Empty training set")
    if verbose:
        print("Binary Search: training on %d files" % len(data))
    # Tìm tham số trên lưới, rồi tinh chỉnh mức dời biên
    params, score = grid_search(data, verbose)
    params, score = refine_trim(data, params, verbose)
    if verbose:
        print("Best parameters:", params, "| training score: %.2f ms" % score)
    return params, score
```

```python
# =====================================================================
# Lớp thuật toán theo giao diện chung của dự án
# =====================================================================
class BinarySearchAlgorithm:
    """Thuật toán 1: Tìm kiếm nhị phân (dựa trên năng lượng)."""

    @property
    def name(self):
        """Tên hiển thị của thuật toán (str)."""
        return "Binary Search"

    @property
    def key(self):
        """Khóa định danh dùng khi lưu tham số/kết quả (str)."""
        return "binary_search"

    def train(self, training_files, training_labs):
        """Huấn luyện trên tập huấn luyện (WAV + LAB).

        Tham số:
            training_files (list[str]): các đường dẫn .wav.
            training_labs (list[str]): các đường dẫn .lab tương ứng.
        Trả về:
            dict: tham số đã huấn luyện, có thể lưu dưới dạng JSON.
        """
        params, _ = train_binary_search(training_files, training_labs)
        return params

    def predict(self, signal, fs, params):
        """Phân đoạn tín hiệu với tham số đã đóng băng (không dùng nhãn kiểm thử).

        Tham số:
            signal (ndarray): tín hiệu cần phân đoạn.
            fs (int): tần số lấy mẫu (Hz).
            params (dict): tham số đã huấn luyện.
        Trả về:
            dict gồm: algorithm_name, speech_mask, segments (giây, giây, 'speech'/'sil'),
            boundaries_ms, feature_times, features, thresholds (T, T_on, T_off), diagnostics.
        """
        # Chạy toàn bộ quy trình phân đoạn và tính thời điểm tâm các khung
        res = predict_file(signal, fs, params)
        feat = res["feat"]
        frame_len, hop = frame_params(fs)
        times = frame_centers(len(feat), frame_len, hop, fs)
        # Đóng gói kết quả (biên đổi sang ms) kèm thông tin phục vụ vẽ đồ thị
        return {
            "algorithm_name": self.name,
            "speech_mask": res["mask"],
            "segments": res["segments"],                       # (giây, giây, 'speech'/'sil')
            "boundaries_ms": [b * 1000.0 for b in res["boundaries"]],
            "feature_times": times,
            "features": {params["kind"] + (" (log10)" if params["use_log"] else ""): feat},
            "thresholds": {"T": params["T"],
                           "T_on": params["T"] + params["hyst"] * params["sep"],
                           "T_off": params["T"] - params["hyst"] * params["sep"]},
            "diagnostics": {"mask_raw": res["mask_raw"], "mask_exp": res["mask_exp"],
                            "xn": res["xn"], "params": dict(params)},
        }
```

### Trained parameters (frozen)

## 5. Testing and evaluation (MAE / RMSE in ms)

```python
def match_boundaries(detected_boundaries_s, gt_boundaries_s):
    """Ghép biên dự đoán với biên chuẩn theo thứ tự thời gian.

    Cách ghép: với mỗi biên chuẩn, chọn biên dự đoán gần nhất chưa được dùng.
    Cách này xử lý được trường hợp số biên dự đoán khác số biên chuẩn.

    Tham số:
        detected_boundaries_s (list[float]): biên dự đoán (giây).
        gt_boundaries_s (list[float]): biên chuẩn (giây).
    Trả về:
        dict gồm: matched_pairs (cặp (chuẩn, dự đoán), giây), errors_ms (sai số tuyệt đối
        mỗi cặp, ms), unmatched_gt (biên chuẩn không có cặp), unmatched_det (biên dự đoán
        không có cặp), num_gt, num_det.
    """
    det = sorted(detected_boundaries_s)
    gt = sorted(gt_boundaries_s)

    matched_pairs = []
    errors_ms = []
    used_det = set()

    # Với mỗi biên chuẩn, tìm biên dự đoán gần nhất chưa ghép
    for gt_b in gt:
        best_idx = None
        best_dist = float('inf')

        for i, det_b in enumerate(det):
            if i in used_det:
                continue
            dist = abs(det_b - gt_b)
            if dist < best_dist:
                best_dist = dist
                best_idx = i

        # Ghi nhận cặp và sai số (đổi sang ms)
        if best_idx is not None:
            used_det.add(best_idx)
            matched_pairs.append((gt_b, det[best_idx]))
            errors_ms.append(abs(det[best_idx] - gt_b) * 1000.0)

    # Các biên chưa được ghép: biên chuẩn bị bỏ sót và biên dự đoán thừa
    unmatched_gt = [gt_b for i, gt_b in enumerate(gt)
                    if not any(gt_b == pair[0] for pair in matched_pairs)]
    unmatched_det = [det[i] for i in range(len(det)) if i not in used_det]

    return {
        'matched_pairs': matched_pairs,
        'errors_ms': errors_ms,
        'unmatched_gt': unmatched_gt,
        'unmatched_det': unmatched_det,
        'num_gt': len(gt),
        'num_det': len(det),
    }


def calculate_mae(errors_ms):
    """Tính sai số tuyệt đối trung bình MAE = mean(|dự đoán - chuẩn|).

    Tham số:
        errors_ms (list[float]): sai số tuyệt đối của các biên (ms).
    Trả về:
        float: MAE (ms); trả về 0.0 nếu danh sách rỗng.
    """
    if len(errors_ms) == 0:
        return 0.0
    return float(np.mean(errors_ms))


def calculate_rmse(errors_ms):
    """Tính căn bậc hai sai số bình phương trung bình RMSE = sqrt(mean(sai_số^2)).

    Tham số:
        errors_ms (list[float]): sai số tuyệt đối của các biên (ms).
    Trả về:
        float: RMSE (ms); trả về 0.0 nếu danh sách rỗng.
    """
    if len(errors_ms) == 0:
        return 0.0
    errors = np.array(errors_ms)
    return float(np.sqrt(np.mean(errors ** 2)))


def evaluate_boundaries(detected_boundaries_s, gt_boundaries_s):
    """Đánh giá đầy đủ biên dự đoán so với biên chuẩn: ghép biên rồi tính MAE, RMSE.

    Tham số:
        detected_boundaries_s (list[float]): biên dự đoán (giây).
        gt_boundaries_s (list[float]): biên chuẩn (giây).
    Trả về:
        dict gồm: mae_ms, rmse_ms, errors_ms, matched_pairs, num_gt, num_det,
        num_matched (số cặp ghép được), num_missing (biên chuẩn bị sót),
        num_extra (biên dự đoán thừa).
    """
    # Ghép biên rồi tính MAE và RMSE trên các cặp đã ghép
    matching = match_boundaries(detected_boundaries_s, gt_boundaries_s)

    mae = calculate_mae(matching['errors_ms'])
    rmse = calculate_rmse(matching['errors_ms'])

    # Tổng hợp kết quả cùng số biên thiếu/thừa
    return {
        'mae_ms': mae,
        'rmse_ms': rmse,
        'errors_ms': matching['errors_ms'],
        'matched_pairs': matching['matched_pairs'],
        'num_gt': matching['num_gt'],
        'num_det': matching['num_det'],
        'num_matched': len(matching['matched_pairs']),
        'num_missing': len(matching['unmatched_gt']),
        'num_extra': len(matching['unmatched_det']),
    }
```

```python
# Nạp các cặp wav + lab của tập kiểm thử và khởi tạo thuật toán
test_files, test_labs = find_pairs(TEST_DIR)
algo = BinarySearchAlgorithm()

# Duyệt từng file kiểm thử: phân đoạn bằng tham số đóng băng, so với biên chuẩn
all_err, results = [], {}
print("%-20s %6s %6s %10s %10s" % ("file", "#gt", "#det", "MAE(ms)", "RMSE(ms)"))
for w, l in zip(test_files, test_labs):
    x, fs = read_wav(w)
    gt_b = lab_boundaries(read_lab(l))
    out = algo.predict(x, fs, params)
    det_b = [b / 1000.0 for b in out["boundaries_ms"]]
    # Tính MAE/RMSE của file, gom sai số để tính tổng, lưu kết quả để vẽ đồ thị
    ev = evaluate_boundaries(det_b, gt_b)
    all_err += ev["errors_ms"]
    results[os.path.basename(w)] = (x, fs, gt_b, out)
    print("%-20s %6d %6d %10.2f %10.2f" % (os.path.basename(w)[:20], ev["num_gt"],
          ev["num_det"], ev["mae_ms"], ev["rmse_ms"]))

# In sai số tổng hợp trên toàn bộ biên của tập kiểm thử
print("-" * 56)
print("TOTAL MAE = %.2f ms | RMSE = %.2f ms (%d boundaries)"
      % (calculate_mae(all_err), calculate_rmse(all_err), len(all_err)))
```

## 6. Result plots for the test files

```python
import re                              # đọc giá trị F0mean trong file .lab
from matplotlib.lines import Line2D        # tạo chú giải thủ công cho các đường biên


def read_lab_f0(path):
    """Đọc giá trị F0mean (Hz) từ file .lab (dòng có chứa 'F0mean').

    Tham số:
        path (str): đường dẫn file .lab.
    Trả về:
        float | None: F0 trung bình (Hz); None nếu file không có dòng F0mean.
    """
    with open(path, "r", encoding="utf-8", errors="ignore") as fh:
        for line in fh:
            # Tìm dòng 'F0mean' và lấy số đầu tiên đứng sau nó
            m = re.search(r"f0\s*mean\D*?([0-9]+(?:\.[0-9]+)?)", line, re.IGNORECASE)
            if m:
                return float(m.group(1))
    return None


def short_time_db(xn, fs, win_ms=30.0, hop_ms=10.0):
    """Tính logSTE và logMA (dB) ngắn hạn với cửa sổ win_ms (chỉ dùng để minh họa).

    Tham số:
        xn (ndarray): tín hiệu đã chuẩn hóa biên độ đỉnh.
        fs (int): tần số lấy mẫu (Hz).
        win_ms (float): độ dài cửa sổ (ms).
        hop_ms (float): bước dịch (ms).
    Trả về:
        (times, log_ste, log_ma): thời điểm tâm khung (giây), 10*log10(năng lượng trung bình)
        và 20*log10(biên độ tuyệt đối trung bình), cùng đơn vị dB.
    """
    # Đổi ms sang số mẫu; đệm số 0 nếu tín hiệu ngắn hơn một cửa sổ
    w, h = int(round(win_ms * fs / 1000.0)), int(round(hop_ms * fs / 1000.0))
    if len(xn) < w:
        xn = np.concatenate((xn, np.zeros(w - len(xn))))
    nf = 1 + (len(xn) - w) // h
    starts = np.arange(nf) * h
    ends = starts + w
    # Tổng tích lũy của |x| và x^2 để tính nhanh trung bình trên mỗi cửa sổ
    c1 = np.concatenate(([0.0], np.cumsum(np.abs(xn))))
    c2 = np.concatenate(([0.0], np.cumsum(xn * xn)))
    ma = (c1[ends] - c1[starts]) / w
    ms = np.maximum(c2[ends] - c2[starts], 0.0) / w
    # Đổi sang dB (cộng hằng số nhỏ để tránh log(0))
    times = (starts + w / 2.0) / fs
    return times, 10.0 * np.log10(ms + 1e-12), 20.0 * np.log10(ma + 1e-12)


def estimate_f0_autocorr(xn, fs, speech_mask, mask_times, win_ms=30.0, hop_ms=10.0,
                         f_min=60.0, f_max=400.0, voicing_thr=0.3):
    """Ước lượng F0 bằng tự tương quan trên từng khung 30 ms (chỉ dùng để minh họa).

    Tham số:
        xn (ndarray): tín hiệu đã chuẩn hóa.
        fs (int): tần số lấy mẫu (Hz).
        speech_mask (ndarray bool): nhãn tiếng nói của thuật toán (chỉ ước lượng F0 ở khung nói).
        mask_times (ndarray): thời điểm tâm các khung của speech_mask (giây).
        win_ms, hop_ms (float): độ dài cửa sổ và bước dịch (ms).
        f_min, f_max (float): dải F0 tìm kiếm (Hz).
        voicing_thr (float): ngưỡng r[lag]/r[0] tối thiểu để coi khung là hữu thanh.
    Trả về:
        (times, f0): thời điểm tâm khung (giây) và F0 (Hz); NaN tại khung vô thanh/im lặng.
    """
    # Chuẩn bị cửa sổ, dải lag tương ứng với [f_min, f_max] và mảng kết quả
    w, h = int(round(win_ms * fs / 1000.0)), int(round(hop_ms * fs / 1000.0))
    nf = 1 + (len(xn) - w) // h if len(xn) >= w else 0
    lag_min, lag_max = int(fs / f_max), min(int(fs / f_min), w - 1)
    times = (np.arange(nf) * h + w / 2.0) / fs
    f0 = np.full(nf, np.nan)
    # Với mỗi khung: tự tương quan, lấy đỉnh trong dải lag, nhận nếu đủ mạnh (hữu thanh)
    for i in range(nf):
        seg = xn[i * h: i * h + w]
        seg = seg - seg.mean()
        r = np.correlate(seg, seg, mode="full")[w - 1:]
        if r[0] <= 1e-9 or lag_min >= lag_max:
            continue
        lag = lag_min + int(np.argmax(r[lag_min: lag_max + 1]))
        if r[lag] / r[0] >= voicing_thr:
            f0[i] = fs / lag
    # Chỉ giữ F0 ở các khung mà thuật toán phân đoạn cho là tiếng nói
    keep = np.interp(times, mask_times, speech_mask.astype(float)) > 0.5
    f0[~keep] = np.nan
    return times, f0
```

```python
# Vẽ 01 figure cho mỗi file kiểm thử (4 khung): dạng sóng + biên, đặc trưng + ngưỡng + nền tiếng nói,
# logSTE/logMA (30 ms), F0 tự tương quan (30 ms) so với F0mean trong file .lab
FIG_DIR = "results"
os.makedirs(FIG_DIR, exist_ok=True)
lab_of = {os.path.basename(w): l for w, l in zip(test_files, test_labs)}   # tên file -> file .lab

for name, (x, fs, gt_b, out) in results.items():
    dur = len(x) / fs
    det_b = [b / 1000.0 for b in out["boundaries_ms"]]
    ev = evaluate_boundaries(det_b, gt_b)
    t_feat, mask = out["feature_times"], out["speech_mask"]
    key = list(out["features"])[0]

    # Dựng lưới 3 hàng: hàng 1 và 2 chiếm cả chiều ngang, hàng 3 chia đôi (logSTE/logMA | F0)
    fig = plt.figure(figsize=(14, 8.5))
    gs = fig.add_gridspec(3, 2, height_ratios=[1.0, 1.5, 1.2])
    ax1 = fig.add_subplot(gs[0, :])
    ax2 = fig.add_subplot(gs[1, :], sharex=ax1)
    ax3 = fig.add_subplot(gs[2, 0], sharex=ax1)
    ax4 = fig.add_subplot(gs[2, 1], sharex=ax1)
    fig.suptitle("Xanh: biên dự đoán · Đỏ nét đứt: biên chuẩn · Nền xanh nhạt: Speech", fontsize=10)

    # Khung 1: dạng sóng với biên chuẩn (đỏ nét đứt) và biên dự đoán (xanh)
    ax1.plot(np.arange(len(x)) / fs, x, lw=0.5, color="gray")
    for b in gt_b:
        ax1.axvline(b, color="red", ls="--", lw=1.3)
    for b in det_b:
        ax1.axvline(b, color="blue", ls="-", lw=1.3)
    ax1.set_title("%s — Dạng sóng và biên chuẩn" % name, fontsize=12)
    ax1.set_ylabel("Biên độ")
    ax1.legend(handles=[Line2D([0], [0], color="red", ls="--", label="biên chuẩn"),
                        Line2D([0], [0], color="blue", ls="-", label="biên dự đoán")],
               loc="upper right", fontsize=8)

    # Khung 2: đặc trưng làm mượt, nền xanh nhạt ở vùng tiếng nói, ngưỡng T, biên chuẩn và biên dự đoán
    ax2.fill_between(t_feat, 0, 1, where=mask, step="mid", color="tab:green", alpha=0.15,
                     transform=ax2.get_xaxis_transform())
    ax2.plot(t_feat, out["features"][key], color="black", lw=0.9)
    ax2.axhline(out["thresholds"]["T"], color="black", lw=0.8)
    ax2.text(1.0, out["thresholds"]["T"], "T=%.5f" % out["thresholds"]["T"], ha="right", va="bottom",
             transform=ax2.get_yaxis_transform(), fontsize=8)
    for b in gt_b:
        ax2.axvline(b, color="red", ls="--", lw=1.3)
    for b in det_b:
        ax2.axvline(b, color="blue", ls="-", lw=1.3)
    # Tiêu đề hai dòng: tên thuật toán; MAE, RMSE và số biên ghép/thừa/thiếu của file
    ax2.set_title("Tìm kiếm nhị phân\nMAE: %.1f ms | RMSE: %.1f ms; ghép/thừa/thiếu %d/%d/%d"
                  % (ev["mae_ms"], ev["rmse_ms"], ev["num_matched"], ev["num_extra"], ev["num_missing"]),
                  fontsize=10)
    ax2.set_ylabel("%s (làm mượt)" % key)

    # Khung 3: logSTE và logMA (dB) với cửa sổ 30 ms trên tín hiệu đã chuẩn hóa
    xn = out["diagnostics"]["xn"]
    t30, log_ste, log_ma = short_time_db(xn, fs)
    ax3.plot(t30, log_ste, color="tab:green", lw=1.0, label="logSTE")
    ax3.plot(t30, log_ma, color="tab:cyan", lw=1.0, label="logMA")
    ax3.set_title("Đặc trưng log (30 ms)", fontsize=11)
    ax3.set_ylabel("Mức (dB)")
    ax3.legend(loc="upper right", fontsize=8)

    # Khung 4: F0 ước lượng bằng tự tương quan (30 ms) và đường F0mean đọc từ file .lab
    t_f0, f0 = estimate_f0_autocorr(xn, fs, mask, t_feat)
    ax4.plot(t_f0, f0, ".-", ms=3, lw=0.8, color="tab:orange", label="F0 ước lượng")
    f0_lab = read_lab_f0(lab_of[name])
    if f0_lab is not None:
        ax4.axhline(f0_lab, color="gray", ls="--", lw=1.2, label="F0mean LAB")
    ax4.set_ylim(0, 450)
    ax4.set_title("F0 tự tương quan (30 ms)", fontsize=11)
    ax4.set_ylabel("F0 (Hz)")
    ax4.legend(loc="upper right", fontsize=8)

    # Định dạng chung: trục thời gian trọn độ dài file, rồi lưu PNG và hiển thị
    for ax in (ax1, ax2, ax3, ax4):
        ax.set_xlabel("Thời gian (giây)", fontsize=9)
    ax1.set_xlim(0, dur)
    fig.tight_layout(rect=[0, 0, 1, 0.96])
    fig.savefig(os.path.join(FIG_DIR, "figure_%s.png" % os.path.splitext(name)[0]), dpi=200)
    plt.show()
```

```python
import csv   # ghi bảng kết quả ra file CSV

# Thư mục lưu bảng (CSV) và hình (PNG) kết quả trung gian
OUT_DIR = "results"
os.makedirs(OUT_DIR, exist_ok=True)


def save_table(header, rows, filename, title=None, max_print=None):
    """Lưu bảng ra file CSV trong OUT_DIR và in bảng căn lề ra màn hình.

    Tham số:
        header (list[str]): tên các cột.
        rows (list[list]): các hàng dữ liệu (mỗi hàng cùng độ dài với header).
        filename (str): tên file CSV (không kèm thư mục).
        title (str | None): tiêu đề in phía trên bảng.
        max_print (int | None): số hàng tối đa in ra màn hình (None = in hết; CSV luôn đủ).
    Trả về:
        str: đường dẫn file CSV đã lưu.
    """
    # Ghi toàn bộ bảng ra CSV (utf-8-sig để Excel đọc đúng tiếng Việt)
    path = os.path.join(OUT_DIR, filename)
    with open(path, "w", newline="", encoding="utf-8-sig") as fh:
        w = csv.writer(fh)
        w.writerow(header)
        w.writerows(rows)

    # Định dạng số thực 3 chữ số thập phân và tính độ rộng từng cột để căn lề
    shown = rows if max_print is None else rows[:max_print]
    cells = [[("%.3f" % v if isinstance(v, float) else str(v)) for v in r] for r in shown]
    width = [max([len(str(h))] + [len(r[i]) for r in cells]) for i, h in enumerate(header)]

    # In tiêu đề, dòng tên cột và các hàng dữ liệu
    if title:
        print("\n" + title)
    print("  ".join(str(h).ljust(wd) for h, wd in zip(header, width)))
    print("  ".join("-" * wd for wd in width))
    for r in cells:
        print("  ".join(v.ljust(wd) for v, wd in zip(r, width)))
    if max_print is not None and len(rows) > max_print:
        print("... (%d hàng, xem đầy đủ trong file CSV)" % len(rows))
    print("-> đã lưu:", path)
    return path


def save_figure(fig, filename):
    """Lưu một figure matplotlib thành file PNG trong OUT_DIR.

    Tham số:
        fig (matplotlib.figure.Figure): figure cần lưu.
        filename (str): tên file PNG (không kèm thư mục).
    Trả về:
        str: đường dẫn file PNG đã lưu.
    """
    path = os.path.join(OUT_DIR, filename)
    fig.savefig(path, dpi=200, bbox_inches="tight")
    print("-> đã lưu:", path)
    return path


def frame_metrics(pred, gt):
    """Tính các chỉ số phân loại mức khung cho lớp tiếng nói (True) so với nhãn chuẩn.

    Tham số:
        pred (ndarray bool): nhãn dự đoán của từng khung.
        gt (ndarray bool): nhãn chuẩn của từng khung (cùng độ dài với pred).
    Trả về:
        dict gồm: n (số khung), acc (độ chính xác), prec (precision), rec (recall), f1.
    """
    # Đếm các ô của ma trận nhầm lẫn (True = tiếng nói)
    tp = int(np.sum(pred & gt))
    fp = int(np.sum(pred & ~gt))
    fn = int(np.sum(~pred & gt))
    tn = int(np.sum(~pred & ~gt))
    # Tính chỉ số, tránh chia cho 0 khi một lớp rỗng
    n = tp + fp + fn + tn
    acc = (tp + tn) / n if n else 0.0
    prec = tp / (tp + fp) if (tp + fp) else 0.0
    rec = tp / (tp + fn) if (tp + fn) else 0.0
    f1 = 2 * prec * rec / (prec + rec) if (prec + rec) else 0.0
    return {"n": n, "acc": acc, "prec": prec, "rec": rec, "f1": f1,
            "tp": tp, "fp": fp, "fn": fn, "tn": tn}
```

```python
# ---------- 8.1 Bảng kết quả trên tập kiểm thử ----------
lab_of = {os.path.basename(w): l for w, l in zip(test_files, test_labs)}   # tên file -> file .lab

# Bảng 1: sai số TỪNG biên (có dấu = biên dự đoán - biên chuẩn; dương = dự đoán muộn hơn)
rows1, rows2, rows3 = [], [], []
signed_all, tot = [], {"gt": 0, "det": 0, "miss": 0, "extra": 0}
pooled = {"tp": 0, "fp": 0, "fn": 0, "tn": 0}
for name, (x, fs, gt_b, out) in results.items():
    det_b = [b / 1000.0 for b in out["boundaries_ms"]]
    ev = evaluate_boundaries(det_b, gt_b)
    # Mỗi cặp biên đã ghép cho một hàng của bảng 1
    signed = []
    for k, (g_t, d_t) in enumerate(ev["matched_pairs"], start=1):
        s = (d_t - g_t) * 1000.0
        signed.append(s)
        rows1.append([name, k, g_t * 1000.0, d_t * 1000.0, s, abs(s)])
    signed_all += signed
    # Bảng 2: tổng hợp theo file (số biên, thiếu/thừa, MAE, RMSE, độ lệch trung bình có dấu)
    bias = float(np.mean(signed)) if signed else 0.0
    rows2.append([name, ev["num_gt"], ev["num_det"], ev["num_missing"], ev["num_extra"],
                  ev["mae_ms"], ev["rmse_ms"], bias])
    for key, val in (("gt", ev["num_gt"]), ("det", ev["num_det"]),
                     ("miss", ev["num_missing"]), ("extra", ev["num_extra"])):
        tot[key] += val
    # Bảng 3: chỉ số mức khung so với nhãn chuẩn tại tâm khung
    gt_mask = lab_frame_mask(read_lab(lab_of[name]), out["feature_times"])
    fm = frame_metrics(out["speech_mask"], gt_mask)
    for key in pooled:
        pooled[key] += fm[key]
    rows3.append([name, fm["n"], fm["acc"], fm["prec"], fm["rec"], fm["f1"]])

# Dòng TỔNG cho bảng 2 (MAE/RMSE tính trên mọi biên) và bảng 3 (gộp số đếm mọi khung)
abs_all = [abs(s) for s in signed_all]
rows2.append(["TOTAL", tot["gt"], tot["det"], tot["miss"], tot["extra"],
              calculate_mae(abs_all), calculate_rmse(abs_all),
              float(np.mean(signed_all)) if signed_all else 0.0])
pm = frame_metrics(np.array([True] * pooled["tp"] + [True] * pooled["fp"]
                            + [False] * pooled["fn"] + [False] * pooled["tn"]),
                   np.array([True] * pooled["tp"] + [False] * pooled["fp"]
                            + [True] * pooled["fn"] + [False] * pooled["tn"]))
rows3.append(["TOTAL", pm["n"], pm["acc"], pm["prec"], pm["rec"], pm["f1"]])

# Xuất ba bảng ra màn hình và file CSV
save_table(["file", "boundary#", "GT (ms)", "Detected (ms)", "Error det-gt (ms)", "|Error| (ms)"],
           rows1, "table1_boundary_errors.csv", "Bảng 1. Sai số từng biên (tập kiểm thử)")
save_table(["file", "#gt", "#det", "missing", "extra", "MAE (ms)", "RMSE (ms)", "mean error (ms)"],
           rows2, "table2_file_summary.csv", "Bảng 2. Tổng hợp theo file")
save_table(["file", "#frames", "accuracy", "precision", "recall", "F1"],
           rows3, "table3_frame_level.csv", "Bảng 3. Chỉ số mức khung (lớp tiếng nói)")
```

```python
# ---------- 8.2 Hình kết quả trung gian trên tập kiểm thử ----------

# Hình 1: sai số có dấu của từng biên (dương = biên dự đoán muộn hơn biên chuẩn)
labels = ["%s #%d" % (r[0].replace(".wav", ""), r[1]) for r in rows1]
errs = [r[4] for r in rows1]
fig, ax = plt.subplots(figsize=(10, 4.5))
bars = ax.bar(range(len(errs)), errs, color=["tab:red" if e > 0 else "tab:blue" for e in errs])
# Ghi giá trị lên từng cột, vẽ đường 0 và đường độ lệch trung bình
for i, e in enumerate(errs):
    ax.text(i, e + (1.0 if e >= 0 else -1.0), "%.1f" % e, ha="center",
            va="bottom" if e >= 0 else "top", fontsize=9)
ax.axhline(0, color="black", lw=1)
ax.axhline(np.mean(errs), color="gray", ls="--", label="Mean error = %.1f ms" % np.mean(errs))
# Đặt nhãn trục, tiêu đề (kèm MAE/RMSE tổng) và lưu hình
ax.set_xticks(range(len(errs)))
ax.set_xticklabels(labels, rotation=45, ha="right")
ax.set_ylabel("Error = detected - ground truth (ms)")
ax.set_title("Boundary errors on test set | MAE = %.2f ms, RMSE = %.2f ms"
             % (calculate_mae(abs_all), calculate_rmse(abs_all)))
ax.legend()
plt.tight_layout()
save_figure(fig, "fig1_boundary_errors.png")
plt.show()

# Hình 2: nhãn tiếng nói qua từng giai đoạn xử lý (chuẩn -> ngưỡng -> dời biên/lọc -> lấp im lặng)
stages = [("Ground truth", None), ("1. Threshold (mask_raw)", "mask_raw"),
          ("2. Trim + min speech (mask_exp)", "mask_exp"), ("3. Fill silence < 200 ms (final)", "speech_mask")]
fig, axes = plt.subplots(len(results), 1, figsize=(12, 2.6 * len(results)), sharex=False)
axes = np.atleast_1d(axes)
for ax, (name, (x, fs, gt_b, out)) in zip(axes, results.items()):
    t = out["feature_times"]
    gt_mask = lab_frame_mask(read_lab(lab_of[name]), t)
    # Mỗi giai đoạn là một dải màu ở một mức y riêng (xếp từ trên xuống dưới)
    for row, (label, key) in enumerate(stages):
        mask = gt_mask if key is None else (out["speech_mask"] if key == "speech_mask"
                                            else out["diagnostics"][key])
        y0 = len(stages) - 1 - row
        ax.fill_between(t, y0, y0 + 0.8, where=mask, step="mid",
                        color="tab:red" if key is None else "tab:blue", alpha=0.7)
    # Nhãn trục y theo tên giai đoạn, tiêu đề theo tên file
    ax.set_yticks([len(stages) - 1 - i + 0.4 for i in range(len(stages))])
    ax.set_yticklabels([s[0] for s in stages], fontsize=8)
    ax.set_title(name, fontsize=10)
    ax.set_xlabel("Time (s)")
    ax.set_xlim(0, len(x) / fs)          # hiển thị trọn độ dài file (kể cả phần im lặng hai đầu)
plt.tight_layout()
save_figure(fig, "fig2_mask_stages.png")
plt.show()
```

```python
# ---------- 8.3 Quá trình tìm kiếm nhị phân trên dữ liệu huấn luyện ----------

def binary_search_trace(f, g, lo, hi, eps_rel=1e-4, max_iter=60, patience=3):
    """Giống binary_search_threshold nhưng ghi lại kết quả từng vòng lặp để trình bày.

    Tham số:
        f, g (ndarray): đặc trưng tiếng nói / im lặng trong vùng chồng lấn.
        lo, hi (float): hai đầu đoạn tìm kiếm.
        eps_rel, max_iter, patience: như trong binary_search_threshold.
    Trả về:
        (T_final, history): T_final là ngưỡng cuối; history là list các bộ
        (vòng, T, F(T), t_min, t_max) với t_min/t_max là đoạn SAU khi cập nhật ở vòng đó.
    """
    # Khởi tạo đoạn tìm kiếm và ngưỡng đầu tiên ở giữa đoạn
    t_min, t_max = lo, hi
    eps = eps_rel * (hi - lo)
    T = 0.5 * (t_min + t_max)
    last, same, history = None, 0, []
    for it in range(1, max_iter + 1):
        # Tính F(T), thu hẹp đoạn tìm kiếm rồi ghi lại trạng thái của vòng này
        F, counts = confusion_error(T, f, g)
        if F > 0:
            t_min = T
        else:
            t_max = T
        history.append((it, T, F, t_min, t_max))
        # Điều kiện dừng giống hệt binary_search_threshold
        if t_max - t_min < eps:
            break
        same = same + 1 if counts == last else 0
        last = counts
        if same >= patience:
            break
        T = 0.5 * (t_min + t_max)
    return 0.5 * (t_min + t_max), history


# Nạp dữ liệu huấn luyện và dựng lại S (khung tiếng nói), Q (khung im lặng) với tham số đang dùng
data_tr = None
if os.path.isdir(TRAIN_DIR):
    train_files, train_labs = find_pairs(TRAIN_DIR)
    data_tr = load_dataset(train_files, train_labs)
    tracks = [compute_feature_track(d["xn"], d["fs"], params["kind"], params["use_log"],
                                    params["smooth_len"]) for d in data_tr]
    S, Q = pooled_class_features(data_tr, tracks)
    lo, hi = find_overlap_region(S, Q)
    f = S[(S >= lo) & (S <= hi)]
    g = Q[(Q >= lo) & (Q <= hi)]

if data_tr is None:
    print("Không có dữ liệu huấn luyện trong '%s': bỏ qua mục 8.3 và 8.4." % TRAIN_DIR)
elif lo >= hi or len(f) == 0 or len(g) == 0:
    print("Hai lớp không chồng lấn: ngưỡng lấy tại trung điểm, không có vòng tìm kiếm để vẽ.")
else:
    # Chạy tìm kiếm nhị phân có ghi vết và đối chiếu với T đang dùng trong PARAMS
    T_trace, hist = binary_search_trace(f, g, lo, hi)
    print("Vùng chồng lấn [%.5f, %.5f] | #speech=%d, #silence=%d khung trong vùng"
          % (lo, hi, len(f), len(g)))
    print("T (truy vết) = %.6f | T (PARAMS) = %.6f | chênh lệch = %.2e"
          % (T_trace, params["T"], abs(T_trace - params["T"])))

    # Bảng 4: từng vòng lặp (vòng, T, F(T), đoạn tìm kiếm, độ rộng đoạn)
    rows4 = [[it, T, F, a, b, b - a] for it, T, F, a, b in hist]
    save_table(["iter", "T", "F(T)", "t_min", "t_max", "width"], rows4,
               "table4_binary_search_iterations.csv", "Bảng 4. Từng vòng tìm kiếm nhị phân")

    # Hình 3a, 3b: T và đoạn tìm kiếm theo vòng lặp; F(T) hội tụ về 0
    its = [h[0] for h in hist]
    fig, axes = plt.subplots(1, 3, figsize=(17, 4.6))
    axes[0].plot(its, [h[1] for h in hist], "o-", color="tab:red", label="T")
    axes[0].fill_between(its, [h[3] for h in hist], [h[4] for h in hist], alpha=0.25, label="[t_min, t_max]")
    axes[0].set(xlabel="Iteration", ylabel="Threshold", title="Search interval shrinking")
    axes[0].legend()
    axes[1].plot(its, [h[2] for h in hist], "o-", color="tab:green")
    axes[1].axhline(0, color="black", lw=1)
    axes[1].set(xlabel="Iteration", ylabel="F(T) = FAR - FRR", title="F(T) converging to 0")

    # Hình 3c: FAR(T) và FRR(T) theo ngưỡng; ngưỡng tìm được nằm tại giao điểm
    Ts = np.linspace(lo, hi, 400)
    far = 1.0 - np.searchsorted(np.sort(g), Ts, side="right") / len(g)   # tỷ lệ khung im lặng > T
    frr = np.searchsorted(np.sort(f), Ts, side="right") / len(f)         # tỷ lệ khung nói <= T
    axes[2].plot(Ts, far, color="tab:blue", label="FAR: silence > T")
    axes[2].plot(Ts, frr, color="tab:orange", label="FRR: speech <= T")
    axes[2].axvline(T_trace, color="red", ls="--", label="T = %.4f" % T_trace)
    axes[2].set(xlabel="Threshold T", ylabel="Rate", title="FAR / FRR vs threshold")
    axes[2].legend()
    plt.tight_layout()
    save_figure(fig, "fig3_binary_search.png")
    plt.show()
```

```python
# ---------- 8.4 Lưới siêu tham số và ảnh hưởng của việc dời biên (dữ liệu huấn luyện) ----------

def grid_search_table(data):
    """Chạy lại lưới siêu tham số GRID (trim = 0) và trả về điểm của TỪNG tổ hợp.

    Tham số:
        data (list[dict]): dữ liệu huấn luyện từ load_dataset.
    Trả về:
        list[list]: mỗi hàng là [kind, use_log, smooth_len, hyst, min_speech_ms, T, score_ms].
    """
    rows = []
    # Vòng ngoài: tính đặc trưng và ngưỡng T cho từng (kind, use_log, smooth_len)
    for kind, use_log, sm in itertools.product(GRID["kind"], GRID["use_log"], GRID["smooth_len"]):
        tracks = [compute_feature_track(d["xn"], d["fs"], kind, use_log, sm) for d in data]
        S, Q = pooled_class_features(data, tracks)
        T, sep = train_threshold(S, Q)
        # Vòng trong: đánh giá từng tổ hợp tham số hậu xử lý bằng hàm mục tiêu
        for hy, ms in itertools.product(GRID["hyst"], GRID["min_speech_ms"]):
            p = {"kind": kind, "use_log": use_log, "smooth_len": sm, "T": float(T),
                 "sep": float(sep), "hyst": hy, "trim_on": 0, "trim_off": 0, "min_speech_ms": ms}
            rows.append([kind, use_log, sm, hy, ms, float(T), objective(data, p)])
    return rows


if data_tr is not None:
    # Bảng 5: điểm huấn luyện của mọi tổ hợp, sắp xếp từ tốt đến kém (CSV đủ, màn hình in 10 dòng đầu)
    grid_rows = sorted(grid_search_table(data_tr), key=lambda r: r[-1])
    save_table(["kind", "use_log", "smooth_len", "hyst", "min_speech_ms", "T", "train score (ms)"],
               grid_rows, "table5_grid_search.csv", "Bảng 5. Kết quả lưới siêu tham số (trim = 0)",
               max_print=10)

    # Hình 4: điểm tốt nhất theo độ dài làm mượt, mỗi đường là một cặp (kind, use_log)
    fig, ax = plt.subplots(figsize=(8, 4.8))
    for kind, use_log in itertools.product(GRID["kind"], GRID["use_log"]):
        ys = [min(r[-1] for r in grid_rows if r[0] == kind and r[1] == use_log and r[2] == sm)
              for sm in GRID["smooth_len"]]
        ax.plot(GRID["smooth_len"], ys, "o-", label="%s%s" % (kind, " (log)" if use_log else ""))
    ax.set(xlabel="Smoothing length (frames)", ylabel="Best training score (ms)",
           title="Effect of smoothing length")
    ax.legend()
    plt.tight_layout()
    save_figure(fig, "fig4_grid_smoothing.png")
    plt.show()

    # Bảng 6 + Hình 5: điểm huấn luyện theo (trim_on, trim_off) quanh bộ tham số đang dùng
    ton_list, toff_list = list(TRIM_RANGE), list(TRIM_RANGE)
    score_grid = np.array([[objective(data_tr, dict(params, trim_on=a, trim_off=b))
                            for b in toff_list] for a in ton_list])
    rows6 = [[a, b, float(score_grid[i, j])] for i, a in enumerate(ton_list) for j, b in enumerate(toff_list)]
    save_table(["trim_on", "trim_off", "train score (ms)"], rows6,
               "table6_trim_grid.csv", "Bảng 6. Điểm huấn luyện theo mức dời biên (khung 10 ms)", max_print=10)

    # Bản đồ nhiệt: ghi điểm vào từng ô và đánh dấu mức đang dùng trong PARAMS
    fig, ax = plt.subplots(figsize=(8, 6))
    im = ax.imshow(score_grid, cmap="viridis_r", origin="lower")
    for i in range(len(ton_list)):
        for j in range(len(toff_list)):
            dark = score_grid[i, j] > np.median(score_grid)      # nền tối -> chữ trắng, nền sáng -> chữ đen
            ax.text(j, i, "%.0f" % score_grid[i, j], ha="center", va="center", fontsize=8,
                    color="white" if dark else "black")
    ax.set_xticks(range(len(toff_list)))
    ax.set_xticklabels(toff_list)
    ax.set_yticks(range(len(ton_list)))
    ax.set_yticklabels(ton_list)
    cur_i, cur_j = ton_list.index(params.get("trim_on", 0)), toff_list.index(params.get("trim_off", 0))
    ax.plot(cur_j, cur_i, "r*", markersize=18, label="PARAMS in use")
    ax.set(xlabel="trim_off (frames)", ylabel="trim_on (frames)", title="Training score (ms) vs boundary trimming")
    fig.colorbar(im, label="score (ms)")
    ax.legend(loc="upper center", bbox_to_anchor=(0.5, -0.1))    # đặt chú giải bên dưới để không che ô
    plt.tight_layout()
    save_figure(fig, "fig5_trim_heatmap.png")
    plt.show()
    print("Điểm huấn luyện của PARAMS đang dùng: %.2f ms" % objective(data_tr, params))
```
